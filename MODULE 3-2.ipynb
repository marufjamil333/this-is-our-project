```python
# Map

numbers = [1,2,3,4,5]
squred_numbers =[]
for num in numbers:
    squred_numbers.append(num**2)
print(squred_numbers)
    
```

```python
#map

numbers = [1,2,3,4,5]
squred_numbers = [num**2 for num in numbers] # list comprehension
print(squred_numbers)
```

```python
#map
price_list = { "Rice" : 60, "Salt": 30, "Sugar":100}
discount_price_list ={}
for item, price in price_list.items():
    discount_price_list[item] = price * 0.9

print(discount_price_list)
```

```python
price_list = { "Rice" : 60, "Salt": 30, "Sugar":100}

discount_price_list = {item:price * 0.9 for item, price in price_list.items()}

print(discount_price_list)
```

```python

# filter

numbers =list (range(1,21))

odd_numbers= []
for num in numbers:
    if num % 2 == 1:
        odd_numbers.append(num)

print(odd_numbers)
```

```python
# comprehension
numbers =list (range(1,21))
odd_numbers = {num for num in numbers if num % 2 ==1}
print(odd_numbers)
```

```python
#filter 
product_list ={"Bread ":False , "Milk":True ,"Juice":False,"Biscuits":True}

new_product_list ={}
for product,is_date_over in product_list.items():
    if not is_date_over:
        new_product_list[product] = is_date_over



new_product_list
```

```python
# Website show all the users information

users = {

"Alvi":{"city":"Dhaka","age":34,"phone":213457},
"Sadia":{"city":"New york ","age":22,"phone":12763},
"samia":{"city":"Rajshahi","age":14,"phone":169754},
"jahid":{"city":"London","age":20,"phone":7645230},
"Hasan":{"city":"Istambul","age":23,"phone":734322}

}

for name, info in users.items():
    #Showing Names
    print("Name:",name,end = " | ")
    #showing information
    for info_key,info_value in info.items():
        print(f"{info_key}:{info_value}", end = " | ")

```

```python
#website - hide sensitive informations (e.g number) for underage users

for name , info in users.items():
    print("Names:",name, end = " | ")
    # showing Information
    if info["age"]>=18:
        #show all informarion
        for info_key,info_value in info.items():
            print(f"{info_key} : {info_value}", end ="  |  " )
    else:
        # Show Information without phone
      for info_key,info_value in info.items(): 
          if not info_key == "phone":
              print(f"{info_key} : {info_value}", end ="  |  " )    
              

    print()          

```

```python
#mini problem per product price + total price calcualtion
products = ["Rice", "Salt", "Sugar"]
quantity = [100, 80, 50]
price_list = [60.5, 30.5, 40.0]
total = 0

for item,qty,price in zip(products,quantity,price_list):
    print(f"Item: {item}, cost: {qty*price}")
    total =total + qty* price

print(f"total cost: {total}")    
```

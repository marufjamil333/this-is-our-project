```python
class Car:
    def __init__(self, brand, color, model, year):
        self.brand = brand
        self.color = color
        self.model = model  #attribute
        self.year = year

    def start(self):
        print(f"{self.brand} {self.model} is starting.")

    def stop(self):
        print(f"{self.brand} {self.model} is stopping.") 

    def drive(self):
        print(f"{self.brand} is driving.")    


my_car = Car("Toyota", "Red", "Camry", 2022)
car2 = Car("Honda", "Blue", "Civic", 2020)
print(my_car.brand)  # Output: Toyota
print(my_car.color)  # Output: Red
my_car.drive()  # Output: Toyota is driving.              
my_car.start()  # Output: Toyota Camry is starting.
my_car.stop()  # Output: Toyota Camry is stopping.

print(car2.brand)  # Output: Honda
print(car2.color)  # Output: Blue   
car2.drive()  # Output: Honda is driving.
car2.start()  # Output: Honda Civic is starting.
car2.stop()  # Output: Honda Civic is stopping.
```

10. Four Pillars of OOP

English:
OOP is built on four main ideas:

Encapsulation
Inheritance
Polymorphism
Abstraction

```python
#Incapsualatiion in oop
class BankAccount:
    def __init__(self,balance):
        self.__balance = balance # Private attribute
    def deposit(self, amount):
        self.__balance += amount

    def get_balance(self):
        return self.__balance    


account = BankAccount(1000)
account.deposit(500)
print(account.get_balance())  # Output: 1500
#print(account.__balance)  # This will raise an AttributeError because __balance is private
```

```python
#Inheritance in OOP
class animal:
    def eat(self):
        print("Animal is eating.")


class Dog(animal):
    def bark(self):
        print("Dog is barking")


d = Dog()
d.eat()  # Output: Animal is eating.
d.bark()  # Output: Dog is barking  
   
```

```python
# Example of polymorphism in OOP

class Cat:
    def sound(self):
        print("Meow")


class Dog:
    def sound(self):
        print("Woof")

# Creating objects of the classes
cat = Cat()
dog = Dog()

# Calling the sound method (polymorphism in action)
cat.sound()  # Output: Meow
dog.sound()  # Output: Woof
```

```python
from abc import ABC,abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass


class Rectangle(Shape):
    def __init__(self,width,height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height


r = Rectangle(5, 10)
print("Area of rectangle:", r.area())  # Output: Area of rectangle: 50

```

```python
class Student:
    def __init__(self, name, roll, marks):
        self.name = name
        self.roll = roll
        self.__marks = marks            # encapsulation

    def get_marks(self):
        return self.__marks

    def save_to_file(self):
        with open("students.txt", "a") as file:
            file.write(f"{self.name},{self.roll},{self.__marks}\n")


def show_all_students():
    try:
        with open("students.txt", "r") as file:
            for line in file:
                name, roll, marks = line.strip().split(",")
                print(f"Name: {name} | Roll: {roll} | Marks: {marks}")
    except FileNotFoundError:
        print("No records found!")


# Using the program
s1 = Student("Rahim", 101, 85)
s2 = Student("Karim", 102, 90)

s1.save_to_file()
s2.save_to_file()

show_all_students()
```

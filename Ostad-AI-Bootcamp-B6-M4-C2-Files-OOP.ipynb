```python
# file handling using contex manager

#instead of manually closing files, use with

#Writing to a file
with open (filepath,"w") as file:
    f.write("hello world")


#reading from a file
with open (filepath,"r") as file:
    content = file.read()   
    print(content)


#appending to file
with open (filepath,"a") as file:
    file.write("new content")
    print(content)
    
```

```python
old_subscriber = {"rafin@mail.com","shafin@mail.com","sadia@mail.com"}
new_subscriber = {"sadia@mail.com","hasan@gmail.com","sayem@mail.com"}

#Total Unique Subscriber
total_unique = old_subscriber.union(new_subscriber)

print("Total Unique Subscriber: ", total_unique)

#User who Shifted to new channel(common subscriber)
shifted_users = old_subscriber.intersection(new_subscriber)
print("Users Shifted to new channel:", shifted_users)

#Users who stayed only in old channel
only_old = old_subscriber.difference(new_subscriber)
print("Only Old Channel:",only_old)

#users who are completely new
only_new = new_subscriber.difference(old_subscriber)
print("only new channel:",only_new)

```

```python
import timeit

n = 1_000_000
lst = list(range(n))
tpl = tuple(range(n))
st  = set(range(n))

target = 999_999   # শেষ এলিমেন্ট (list/tuple-এর জন্য worst case)

list_time  = timeit.timeit("target in lst", globals=globals(), number=1000)
tuple_time = timeit.timeit("target in tpl", globals=globals(), number=1000)
set_time   = timeit.timeit("target in st",  globals=globals(), number=1000)

print(f"List : {list_time:.4f} seconds")
print(f"Tuple: {tuple_time:.4f} seconds")
print(f"Set  : {set_time * 1_000_000:.2f} microseconds (১০০০ বারের মোট)")
```

```python
# A dictionary works like a search engine.
# you give a key, and it instantly returns the value

student_Info = {
    "maruf":{"roll":566,"city":"dhaka","age":56},
    "Puspita":{"roll":553,"city":"rajshai","age":35}
}

#get full info
print(student_Info.get("maruf"))

# Get nested value safely
maruf_Info = student_Info.get("maruf")
print(maruf_Info.get("city"))

# Using defualt value if key does not exist
print(student_Info.get("Hasen", "sudent NOt  found "))

```

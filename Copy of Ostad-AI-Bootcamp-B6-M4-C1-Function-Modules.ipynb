```python
# So We Use Funtion this the good practice

def calculate_final_price(price):
    tax_rate = 0.25
    total_price = price + (price * tax_rate)
    print(f"final price is : {total_price}")
```

```python
import random

print(random.randint(1, 10))          # ১ থেকে ১০ এর মধ্যে র‍্যান্ডম পূর্ণসংখ্যা (দুটোই অন্তর্ভুক্ত)
print(random.random())                # ০.০ থেকে ১.০ এর মধ্যে র‍্যান্ডম দশমিক সংখ্যা
print(random.uniform(1.5, 5.5))       # নির্দিষ্ট রেঞ্জে র‍্যান্ডম দশমিক সংখ্যা

fruits = ["apple", "banana", "mango"]
print(random.choice(fruits))          # একটি র‍্যান্ডম আইটেম বাছাই
print(random.sample(fruits, 2))       # ২টি ইউনিক র‍্যান্ডম আইটেম বাছাই

random.shuffle(fruits)                # list নিজেই এলোমেলো হয়
print(fruits)
```

```python
#datetime 
from datetime import datetime,date,timedelta
now = datetime.now()
print(now)

today = date.today()
print(today)

print(now.year,now.month,now.day)
print(now.strftime("%d-%m-%Y %H:%M"))
tomorrow = today + timedelta(days=1)  # তারিখের যোগ-বিয়োগ
print(tomorrow)
```

```python
#scenario:An E- commerce system to calculate final price of product

#Case 1 
price = 1000
tax_rate = 0.15 #0.25%
total_price = price + (price * tax_rate)
print(f"final price : {total_price}")
```

```python
# So We Use Funtion this the good practice

def calculate_final_price(price):
    tax_rate = 0.25
    total_price = price + (price * tax_rate)
    print(f"final price is : {total_price}")
```

```python
# so we use function then advantage is reusable like that machine
calculate_final_price(3000)
calculate_final_price(2480)
calculate_final_price(446)
calculate_final_price(220)
```
